<h2 align="center">RESUMO</h2>

A ampliação do acesso gratuito à internet em espaços públicos expõe usuários a riscos como redes falsas (*Evil Twin*), interceptação de tráfego e páginas fraudulentas de autenticação. Este estudo exploratório, desenvolvido em um projeto de extensão universitária, teve como objetivo reunir e analisar evidências públicas sobre fraudes digitais e comportamento de usuários em redes Wi-Fi públicas e avaliar a cobertura temática de sete objetos de aprendizagem produzidos pelos autores. Adotou-se abordagem quantitativa e documental, com dados secundários do CERT.br, do Fórum Brasileiro de Segurança Pública (FBSP), de pesquisas de fornecedores de segurança e de estudos científicos, tratados em Python com estatística descritiva, intervalos de confiança de Wilson, análise de sensibilidade e análise de poder estatístico. Os resultados indicam que o *phishing* respondeu por 94,96% (22.818 de 24.030) das notificações de fraude ao CERT.br em 2025; que os registros policiais de estelionato por meio eletrônico chegaram a 346.753 em 2025, o equivalente a 15,3% das 2.261.055 ocorrências de estelionato; e que as pesquisas disponíveis apontam persistência de comportamentos de risco em Wi-Fi público, como a crença de que as informações enviadas são seguras (58% dos brasileiros ouvidos em 2017) e o prosseguimento diante de alertas de segurança (70,2% dos alertas SSL do Chrome em 2013). Não foram localizadas medições diretas de golpes originados em Wi-Fi público no Brasil, e as fontes diferem em ano, país e amostra; por isso, as estimativas apresentadas são ordens de grandeza, e não medições. A análise de poder mostrou que um teste de conhecimento antes e depois da leitura do material, com 8 a 15 participantes, só detecta efeitos grandes (tamanho de efeito d<sub>z</sub> de pelo menos 0,8, com 15 participantes), sendo necessários cerca de 34 participantes para um efeito moderado (d<sub>z</sub> = 0,5). Conclui-se que o engano do usuário é o elo comum entre os dados analisados, que HTTPS e VPN não substituem a verificação da rede e que o material produzido cobre os principais riscos identificados, mas seus itens de avaliação apresentam pistas de construção (a alternativa correta é a mais longa em todos os 15 itens) e pré e pós-teste pouco paralelos, devendo a efetividade ser avaliada com amostra maior.

**Palavras-chave:** Wi-Fi público. Segurança da informação. *Evil Twin*. Educação digital. Extensão universitária.

<h2 align="center">ABSTRACT</h2>

The expansion of free internet access in public places exposes users to risks such as rogue access points (Evil Twin), traffic interception and fraudulent login pages. This exploratory study, developed within a university extension project, aimed to gather and analyze public evidence on digital fraud and user behavior on public Wi-Fi networks and to assess the thematic coverage of seven learning objects produced by the authors. A quantitative, document-based approach was adopted, using secondary data from CERT.br, the Brazilian Forum on Public Security (FBSP), security-vendor surveys and scientific studies, processed in Python with descriptive statistics, Wilson confidence intervals, sensitivity analysis and statistical power analysis. Results indicate that phishing accounted for 94.96% (22,818 of 24,030) of the fraud notifications received by CERT.br in 2025; that police records of electronic fraud reached 346,753 in 2025, equivalent to 15.3% of the 2,261,055 fraud occurrences; and that available surveys point to persistent risky behavior on public Wi-Fi, such as the belief that transmitted information is safe (58% of Brazilians surveyed in 2017) and proceeding despite security warnings (70.2% of Chrome SSL warnings in 2013). No direct measurements of scams originating on public Wi-Fi in Brazil were found, and the sources differ in year, country and sample; therefore, the estimates presented are orders of magnitude, not measurements. The power analysis showed that a pre-post knowledge test with 8 to 15 participants detects only large effects (effect size d<sub>z</sub> of at least 0.8 with 15 participants), and about 34 participants are needed for a moderate effect (d<sub>z</sub> = 0.5). It is concluded that user deception is the common thread across the data, that HTTPS and VPN do not replace verifying the network, and that the material covers the main identified risks, but its assessment items show construction cues (the correct option is the longest in all 15 items) and weakly parallel pre and post tests; its effectiveness should be assessed with a larger sample.

**Keywords:** Public Wi-Fi. Information security. Evil Twin. Digital education. University extension.

# 1 INTRODUÇÃO

## 1.1 Contextualização

O acesso gratuito à internet em cafeterias, aeroportos, hotéis, universidades e centros comerciais tornou-se parte da rotina de milhões de pessoas. A oferta desses pontos de acesso cresce de forma expressiva: projeções da Cisco e da Statista citadas por Choi, Carpenter e Ko (2022) estimavam cerca de 94 milhões de *hotspots* públicos no mundo em 2016 e 549 milhões em 2022. Contudo, em redes abertas o usuário não tem como confirmar, apenas pelo nome exibido, quem opera o ponto de acesso. Os identificadores de uma rede (SSID e BSSID) podem ser falsificados com facilidade, de modo que um ponto de acesso falso pode ser indistinguível de um legítimo (LANZE et al., 2015).

No Brasil, o ambiente de fraudes digitais é expressivo. O Anuário Brasileiro de Segurança Pública registrou 2.261.055 ocorrências de estelionato em 2025, das quais 346.753 por meio eletrônico (DINO, 2026). Segundo pesquisa do Fórum Brasileiro de Segurança Pública, 15,8% da população com 16 anos ou mais foi vítima de golpe ou perdeu dinheiro pela internet ou pelo celular no último ano, o equivalente a 26,3 milhões de pessoas (DINO, 2026). Esses números não indicam quantos golpes tiveram origem em Wi-Fi público, mas descrevem o cenário em que o uso dessas redes está inserido.

Este estudo integra o Projeto de Extensão FCI: Segurança Digital (Linha 8: Wi-Fi público e redes sem fio) e parte da mensagem central adotada pelos autores em seus materiais educativos: *Wi-Fi gratuito não significa Wi-Fi confiável* (DUARTE; GOMES, 2026).

## 1.2 Problema de pesquisa e perguntas

**Problema:** que evidências públicas disponíveis permitem dimensionar os riscos associados ao uso de Wi-Fi público no Brasil e fundamentar um material educativo, e em que medida o material produzido pelos autores cobre esses riscos?

**Perguntas específicas:**

- **PE1.** Qual é o panorama de incidentes e fraudes notificados e registrados no Brasil, e qual a participação do *phishing*?
- **PE2.** O que a literatura científica e as pesquisas disponíveis mostram sobre o comportamento de usuários em Wi-Fi público e diante de alertas de segurança?

- **PE4.** Que temas de risco são cobertos pelos objetos de aprendizagem do projeto, e quais ficam descobertos?
- **PE5.** Quantos participantes seriam necessários para avaliar, em um teste antes e depois, a aprendizagem proporcionada pelo material?

## 1.3 Objetivos

**Objetivo geral:** reunir, organizar e analisar evidências públicas sobre fraudes digitais e comportamento de usuários em redes Wi-Fi públicas e avaliar a cobertura temática dos objetos de aprendizagem desenvolvidos no projeto de extensão.

A extensão universitária busca devolver à sociedade conhecimento aplicável. # 2 REFERENCIAL TEÓRICO

## 2.1 Redes sem fio e segurança do acesso

As redes locais sem fio de uso cotidiano baseiam-se na família de padrões IEEE 802.11, que especifica as camadas de controle de acesso ao meio e física (INSTITUTE OF ELECTRICAL AND ELECTRONICS ENGINEERS, 2021). Em redes abertas, o dispositivo associa-se a um ponto de acesso identificado por um nome (SSID) e por um identificador de hardware (BSSID), ambos informados pelo próprio ponto de acesso. Como esses identificadores podem ser falsificados de forma trivial, é possível atrair o tráfego de usuários para pontos de acesso falsos (LANZE et al., 2014).

Além disso, qualquer pessoa dentro do alcance de uma rede 802.11 pode, com *software* gratuito, coletar o tráfego web não criptografado de outros usuários (CONSOLVO et al., 2010). Dispositivos que usam *hotspots* públicos também não têm como impor proteções fortes ao ponto de acesso, o que expõe informações privadas em serviços como e-mail, redes sociais e comércio eletrônico (CHOI; CARPENTER; KO, 2022).

## 2.2 Principais ameaças em redes Wi-Fi públicas

**Rede falsa (*Evil Twin*).** A personificação de pontos de acesso existentes por pontos falsos, para atrair seu tráfego, é conhecida como ataque *Evil Twin* (LANZE et al., 2014). A literatura registra a dificuldade de distinguir, do lado do usuário, um ponto falso de um legítimo (LANZE et al., 2015).

**Interceptação de tráfego e *Man-in-the-Middle*.** A interceptação ocorre quando um terceiro observa a comunicação; no ataque do intermediário, o atacante se coloca entre as partes para observar ou alterar mensagens. Um ponto de acesso controlado pelo atacante é uma posição favorável para esse tipo de ataque, porque o tráfego do usuário passa por ele (LANZE et al., 2014).

**Páginas falsas de autenticação e *phishing*.** Portais de acesso pedem, em geral, aceite de termos ou identificação para liberar a conexão. Páginas falsas imitam esse mecanismo para obter senhas e dados. Os dados do CERT.br mostram que o *phishing* é a forma dominante de fraude notificada no país (seção 4.1), o que torna a verificação do que a página solicita parte essencial da proteção.

**Conexão automática e dispositivos desatualizados.** Dispositivos tendem a se reconectar a redes cujos nomes já conhecem, e sistemas desatualizados podem conter falhas já corrigidas por fabricantes. Essas duas condições são tratadas como fatores de risco no material do projeto (DUARTE; GOMES, 2026), que as converte em orientações práticas (desativar a conexão automática e manter sistema, navegador e aplicativos atualizados).

## 2.3 Criptografia em trânsito e VPN: alcance e limites

O protocolo TLS 1.3 foi projetado para que aplicações cliente e servidor se comuniquem de modo a evitar espionagem, adulteração e falsificação de mensagens (RESCORLA, 2018). O HSTS é um mecanismo pelo qual os sítios declaram que só devem ser acessados por conexões seguras (HODGES; JACKSON; BARTH, 2012). A proteção, porém, refere-se ao canal entre o navegador e o servidor: ela não atesta que o operador do site seja idôneo. Por isso, o cadeado do navegador não deve ser lido como selo de confiança no conteúdo, e o material do projeto reforça essa distinção (DUARTE; GOMES, 2026).

Uma VPN cria um túnel criptografado entre o dispositivo e um servidor VPN. Ela protege o trajeto até esse servidor, mas desloca a confiança para o provedor da VPN e não impede *phishing* nem corrige um aparelho comprometido (DUARTE; GOMES, 2026). Na pesquisa da NordVPN com usuários brasileiros de VPN, 52,4% disseram recorrer a opções gratuitas (VPN..., 2023), o que torna relevante avaliar a confiabilidade do provedor.

## 2.4 Fator humano: percepção de risco e comportamento

Estudos sobre o uso cotidiano de Wi-Fi indicam que os usuários compreendem razoavelmente os aspectos práticos da conexão, mas não os riscos de privacidade, e que práticas de proteção incompletas geram falsa sensação de segurança (KLASNJA et al., 2009). Em experimentos sobre decisões de conexão, a maioria dos participantes declarou que usaria Wi-Fi público em circunstâncias em que o risco assumido não é coerente com a maximização da utilidade, e as mulheres, embora avaliassem os riscos como mais altos, mais frequentemente declararam usar essas redes para economizar o plano de dados (SOMBATRUANG; SASSE; BADDELEY, 2016).

Na observação de redes reais, a maior parte dos usuários evitava acessar sites bancários em redes públicas estabelecidas, mas continuava usando essas redes para redes sociais, e-mail e outros sítios que tratam informações sensíveis (MAIMON et al., 2017). Em *survey* com 1.313 respondentes nos Estados Unidos, fatores como motivação de evitação de ameaças, propensão a evitar riscos e motivações intrínsecas e extrínsecas afetaram a intenção de usar Wi-Fi público; o mesmo trabalho cita que apenas 4,8% das pessoas consideram o Wi-Fi público muito seguro, e ainda assim a maioria o utiliza (CHOI; CARPENTER; KO, 2022).

O comportamento diante de avisos do navegador também é relevante para o caso prático do projeto (continuar mesmo assim após um alerta de certificado). Em um estudo de campo com mais de 25 milhões de alertas observados por telemetria no Chrome e no Firefox, 70,2% dos alertas SSL do Chrome foram ignorados pelos usuários, o que indica que o desenho do alerta influencia o comportamento (AKHAWE; FELT, 2013).

## 2.5 Educação e conscientização em segurança

Há materiais educativos institucionais consolidados em português, como a *Cartilha de Segurança para Internet* do CERT.br e do NIC.br (CERT.BR; NIC.BR, [s.d.]), e orientações recentes do Ministério das Comunicações sobre Wi-Fi público, entre elas confirmar o nome da rede e desconectar-se ao terminar o uso (BRASIL, 2026). O repositório do projeto reúne, como base da Etapa 1, orientações da CISA, da NSA e da FTC e o fascículo sobre redes da Cartilha do CERT.br (CYBERSECURITY AND INFRASTRUCTURE SECURITY AGENCY, [s.d.]; NATIONAL SECURITY AGENCY, 2021; FEDERAL TRADE COMMISSION, [s.d.]; CERT.BR; NIC.BR, [s.d.]). A contribuição específica do projeto é um conjunto curto de objetos de aprendizagem (artigo, guia, caso prático, desafio, quiz, teste final e infográfico) voltado ao público geral, no formato de repositório aberto (DUARTE; GOMES, 2026). A efetividade desses objetos, porém, ainda não foi medida, o que motiva o plano de teste descrito na seção 3.6.

## 2.6 Marco legal brasileiro

O uso de dados pessoais é regulado pela Lei Geral de Proteção de Dados Pessoais (BRASIL, 2018), e os princípios, direitos e deveres do uso da internet constam do Marco Civil da Internet (BRASIL, 2014). A Lei nº 14.155/2021 tornou mais graves os crimes de violação de dispositivo informático, furto e estelionato cometidos de forma eletrônica ou pela internet (BRASIL, 2021). Esses instrumentos delimitam responsabilidades de operadores de rede e controladores de dados, mas não substituem a atenção do usuário final, foco dos materiais deste projeto.

# 3 METODOLOGIA

## 3.1 Classificação da pesquisa

Trata-se de pesquisa aplicada, de natureza exploratória e descritiva, com abordagem predominantemente quantitativa. ## 3.4 Tratamento estatístico

Os cálculos foram feitos em Python (*pandas*, *NumPy*, *SciPy* e *Matplotlib*), com semente aleatória fixa (42) para qualquer procedimento de reamostragem.

**Intervalo de confiança de Wilson** (WILSON, 1927)**.** Para uma proporção observada $\hat p$ em amostra de tamanho $n$, o intervalo de confiança de 95% ($z = 1{,}96$) é:

$$ \frac{\hat p + \frac{z^2}{2n} \pm z\sqrt{\frac{\hat p(1-\hat p)}{n} + \frac{z^2}{4n^2}}}{1 + \frac{z^2}{n}} \tag{1} $$

Foi usado apenas quando o tamanho da amostra é informado e os dados são de amostra (Panda Security, $n = 1.014$).

**Taxa composta de variação anual (CAGR).** Entre um valor inicial $V_i$ e um final $V_f$ separados por $t$ anos:

$$ \mathrm{CAGR} = \left(\frac{V_f}{V_i}\right)^{1/t} - 1 \tag{2} $$

**Inclinação robusta de tendência.** Para a série anual do CERT.br, usou-se o estimador de Theil–Sen (SEN, 1968) (mediana das inclinações entre todos os pares de pontos), menos sensível a valores extremos que a regressão por mínimos quadrados. A inclinação descreve a série, e não prova causa.

**Estimativas do tipo Fermi.** Cada estimativa é o produto de parâmetros declarados (arquivo `estimativas_parametros.csv`), com análise de sensibilidade sobre as premissas e propagação do erro de arredondamento dos dados de entrada.

**Poder estatístico do teste pareado.** Para uma diferença média padronizada $d_z$ (diferença média dividida pelo desvio-padrão das diferenças) e $n$ pares, o poder do teste $t$ bilateral com nível $\alpha$ é calculado pela distribuição $t$ não central, com parâmetro de não centralidade $\delta = d_z\sqrt{n}$:

$$ 1-\beta = 1 - F_{\nu,\delta}(t_{1-\alpha/2,\,\nu}) + F_{\nu,\delta}(-t_{1-\alpha/2,\,\nu}), \qquad \nu = n-1 \tag{3} $$

em que $F_{\nu,\delta}$ é a função de distribuição da $t$ não central. Adotaram-se $\alpha = 0{,}05$ e poder de 80%.

**Análise dos resultados do teste.** Para as pontuações antes e depois, serão calculados a diferença média, o intervalo de confiança por *bootstrap* percentílico (EFRON, 1979) (10.000 reamostragens), o teste dos postos sinalizados de Wilcoxon (WILCOXON, 1945) e o tamanho de efeito $d_z$.

````python
def parse_itens(txt, instrumento):
    partes = re.split(r"^###\s*(\d+)\.\s*", txt, flags=re.M)
    gab = {int(a): b for a, b in re.findall(r"^(\d+)\.\s+\*\*([A-D])\.\*\*", txt, flags=re.M)}
    out = []
    for i in range(1, len(partes), 2):
        n = int(partes[i]); bloco = re.split(r"\n---", partes[i + 1])[0]
        linhas_ = [l.strip() for l in bloco.strip().split("\n") if l.strip()]
        alts = {}
        for l in linhas_[1:]:
            m = re.match(r"^([A-D])\)\s*(.*)$", l)
            if m: alts[m.group(1)] = m.group(2).strip()
        out.append({"instrumento": instrumento, "item": n, "enunciado": linhas_[0].replace("`", ""), "alternativas": alts, "correta": gab.get(n)})
    return out

def tema_principal(it):
    corr = it["alternativas"][it["correta"]]
    texto = sem_acentos(it["enunciado"] + " " + corr + " " + corr)  # a alternativa correta pesa em dobro
    cont = {t: len(re.findall(rx, texto)) for t, rx in TEMAS.items()}
    mx_ = max(cont.values())
    return next((t for t, c in cont.items() if c == mx_), "Outros") if mx_ > 0 else "Outros"

if MATERIAL_OK:
    ITENS = parse_itens(TEXTOS["Quiz"], "Quiz") + parse_itens(TEXTOS["Teste final"], "Teste final")
    reg = []
    for it in ITENS:
        alts = it["alternativas"]; cor = alts[it["correta"]]
        distr = [v for k, v in alts.items() if k != it["correta"]]
        reg.append({"instrumento": it["instrumento"], "item": it["item"], "tema": tema_principal(it), "correta": it["correta"],
                    "chars_correta": len(cor), "chars_media_distratores": float(np.mean([len(v) for v in distr])),
                    "correta_mais_longa": len(cor) == max(len(v) for v in alts.values()),
                    "enunciado": it["enunciado"], "alternativas": " | ".join(f"{k}) {v}" for k, v in alts.items())})
    it_df = pd.DataFrame(reg); it_df.to_csv(DADOS / "itens_avaliacao.csv", index=False, float_format="%.2f")
else:
    it_df = pd.read_csv(DADOS / "itens_avaliacao.csv")

linhas = [[r.instrumento, int(r["item"]), r.tema, r.correta, "sim" if r.correta_mais_longa else "não",
           f"{br(r.chars_correta)} / {br(r.chars_media_distratores, 0)}"] for _, r in it_df.iterrows()]
tabela(11, ["Instrumento", "Item", "Tema principal (classificação automática)", "Correta", "Correta é a mais longa?", "Caracteres (correta / média dos distratores)"], linhas,
       "elaborado pelos autores a partir de DUARTE; GOMES (2026).",
       "a classificação temática é automática, por expressões regulares aplicadas ao enunciado e à alternativa correta, e deve ser revisada manualmente.",
       alinhar=["l", "r", "l", "c", "c", "r"])
k = int(it_df.correta_mais_longa.sum()); n_it = len(it_df)
bt = stats.binomtest(k, n_it, 0.25, alternative="greater")
dist_letras = it_df.correta.value_counts().sort_index().to_dict()
display(Markdown(
    f"- Itens analisados: **{n_it}** (quiz: {int((it_df.instrumento == 'Quiz').sum())}; teste final: {int((it_df.instrumento == 'Teste final').sum())}).\n"
    f"- A alternativa correta é a mais longa em **{k} de {n_it}** itens (**{pct(k / n_it * 100, 0)}**); ao acaso, com quatro alternativas, esperam-se 25%. Teste binomial unilateral: p = **{'< 0,0001' if bt.pvalue < 1e-4 else br(bt.pvalue, 4)}**.\n"
    f"- Distribuição das letras corretas: {', '.join(f'{a} = {b} ({pct(b / n_it * 100, 0)})' for a, b in dist_letras.items())}."
))
````

## 4.6 Teste com pessoas: planejamento, poder estatístico e análise

**Planejamento.** O instrumento do Apêndice B usa cinco itens no pré-teste e cinco no pós-teste, ou seja, pontuações de 0 a 5. A pergunta é quantos participantes são necessários para detectar um ganho de pontuação com o teste $t$ pareado, com $\alpha = 0{,}05$ e poder de 80% (equação 3). O tamanho de efeito é expresso em $d_z$ (diferença média dividida pelo desvio-padrão das diferenças). Por convenção, 0,5 é um efeito moderado e 0,8, grande (COHEN, 1988). Como exemplo ilustrativo, se o desvio-padrão das diferenças fosse de 1 ponto, $d_z = 0{,}8$ corresponderia a um ganho médio de 0,8 ponto em 5.

````python
ALFA, PODER = pv("alfa"), pv("poder_alvo")
def poder_pareado(n, dz, alfa=ALFA):
    gl = n - 1; tc = stats.t.ppf(1 - alfa / 2, gl); nc = dz * math.sqrt(n)
    return 1 - stats.nct.cdf(tc, gl, nc) + stats.nct.cdf(-tc, gl, nc)
def n_necessario(dz, alvo=PODER):
    return next(n for n in range(3, 1000) if poder_pareado(n, dz) >= alvo)

DZS = [0.5, 0.8, 1.0, 1.2]; NS = [8, 12, 15, 20, 30]
linhas = [[br(dz, 1), br(n_necessario(dz)), br(n_necessario(dz, .90))] + [pct(poder_pareado(n, dz) * 100, 0) for n in NS] for dz in DZS]
tabela(12, ["d_z", "n para 80% de poder", "n para 90% de poder"] + [f"Poder com n = {n}" for n in NS], linhas,
       "elaborado pelos autores (cálculo pela distribuição t não central).",
       "teste t pareado bilateral, α = 0,05. Para um instrumento de 5 itens e amostra pequena, o teste de Wilcoxon (previsto na análise) tem poder semelhante, mas esta tabela usa o teste t como aproximação.")

legenda("Gráfico", 12)
fig, ax = plt.subplots(figsize=(7.4, 3.6))
ns = np.arange(4, 61)
estilos = {0.5: ("-", CINZA), 0.8: ("-", PRETO), 1.0: ("--", PRETO), 1.2: (":", PRETO)}
for dz in DZS:
    ls, c = estilos[dz]
    ax.plot(ns, [poder_pareado(n, dz) * 100 for n in ns], ls=ls, color=c, linewidth=1.6, label=f"$d_z$ = {br(dz, 1)}")
ax.axhline(80, color=VERM, linewidth=1); ax.text(60, 81.5, "80%", color=VERM, ha="right", fontsize=8)
ax.axvspan(8, 15, color=CLARO, alpha=.45, label="Plano atual: 8 a 15 participantes")
ax.set_xlabel("Número de participantes (pares pré e pós)"); ax.set_ylabel("Poder estatístico (%)"); ax.set_ylim(0, 102); ax.set_xlim(4, 60)
ax.legend(loc="lower right", fontsize=8)
plt.show(); salvar(fig, "grafico_12_poder_estatistico"); plt.close(fig)
fonte("elaborado pelos autores.")
````

Com 8 a 15 participantes, o poder só é adequado para efeitos grandes: com 15 pares, atinge-se 82% de poder para $d_z$ = 0,8; para um efeito moderado ($d_z$ = 0,5) seriam necessários 34 pares (Tabela 12). O plano de 8 a 15 participantes é suficiente para uma **avaliação de usabilidade** (clareza, utilidade e comentários qualitativos), mas insuficiente para afirmar, com segurança estatística, que o material melhora o desempenho. Recomenda-se, portanto, apontar 34 participantes como meta e tratar resultados com menos pessoas como exploratórios.

````python
ARQ_TESTE = DADOS / "resultados_teste_pessoas.csv"
def analisar_teste(df):
    """Análise pré e pós (pareada) e de percepção. Roda somente com dados reais."""
    d = df[~df.id_participante.astype(str).str.upper().str.startswith("EXEMPLO")].copy()
    pre_c = [f"pre_q{i}" for i in range(1, 6)]; pos_c = [f"pos_q{i}" for i in range(1, 6)]
    d["pre"] = d[pre_c].sum(axis=1); d["pos"] = d[pos_c].sum(axis=1); d["dif"] = d["pos"] - d["pre"]
    n = len(d)
    if n < 2:
        display(Markdown("Dados insuficientes (n < 2) para a análise pareada.")); return
    rng = np.random.default_rng(42)
    boots = np.array([rng.choice(d["dif"].values, size=n, replace=True).mean() for _ in range(10000)])
    ic = np.percentile(boots, [2.5, 97.5])
    sd = d["dif"].std(ddof=1); dz = d["dif"].mean() / sd if sd > 0 else float("nan")
    try: p_w = stats.wilcoxon(d["dif"], zero_method="wilcox").pvalue
    except ValueError: p_w = float("nan")
    def lin(nome, s): return [nome, br(s.mean(), 2), br(s.std(ddof=1), 2), br(s.median(), 1), f"{int(s.min())} a {int(s.max())}"]
    tabela(14, ["Medida", "Média", "Desvio-padrão", "Mediana", "Mínimo a máximo"],
           [lin("Pré-teste (0 a 5)", d["pre"]), lin("Pós-teste (0 a 5)", d["pos"]), lin("Diferença (pós − pré)", d["dif"])],
           "dados coletados pelos autores no teste com pessoas.",
           f"n = {n} participantes. Diferença média = {br(d['dif'].mean(), 2)} (IC 95% bootstrap: {br(ic[0], 2)} a {br(ic[1], 2)}); "
           f"d_z = {br(dz, 2)}; teste de Wilcoxon: p = {br(p_w, 4)}. O delineamento sem grupo de controle não permite atribuir a diferença exclusivamente ao material.",
           titulo="Resultado do teste com pessoas: pontuação antes e depois")
    ln = []
    for col, nome in (("clareza_1a5", "Clareza (1 a 5)"), ("utilidade_1a5", "Utilidade (1 a 5)")):
        s = d[col].dropna()
        if len(s): ln.append([nome, br(s.mean(), 2), br(s.median(), 1), f"{br(s.quantile(.25), 1)} a {br(s.quantile(.75), 1)}"])
    tabela(15, ["Percepção", "Média", "Mediana", "Quartis (Q1 a Q3)"], ln,
           "dados coletados pelos autores no teste com pessoas.",
           "compartilharia: " + ", ".join(f"{k} = {v}" for k, v in d["compartilharia"].value_counts().items()) + ".",
           titulo="Resultado do teste com pessoas: percepção sobre o material")

existe = ARQ_TESTE.exists()
n_reais = 0
if existe:
    df_teste = pd.read_csv(ARQ_TESTE)
    n_reais = int((~df_teste.id_participante.astype(str).str.upper().str.startswith("EXEMPLO")).sum())
sit = "Aguardando coleta de dados" if n_reais == 0 else f"Análise executada com {n_reais} participante(s)"
tabela(13, ["Item", "Situação"], [
    ["Arquivo esperado", "`dados/resultados_teste_pessoas.csv` (modelo em `resultados_teste_pessoas_modelo.csv`)"],
    ["Arquivo existe?", "sim" if existe else "não"],
    ["Participantes com dados reais", br(n_reais)],
    ["Situação", sit],
    ["Meta de participantes", "34 (efeito moderado) ou, no mínimo, 15 (efeito grande)"],
], "elaborado pelos autores.", alinhar=["l", "l"])
if n_reais >= 2:
    analisar_teste(df_teste)
else:
    display(Markdown("A análise pareada, os intervalos de confiança por *bootstrap* e as tabelas de resultado (Tabelas 14 e 15) serão gerados automaticamente "
                     "quando o arquivo `dados/resultados_teste_pessoas.csv` for preenchido com dados reais."))
````

## 4.7 Discussão integrada

**PE1 – panorama.** Os registros policiais de estelionato cresceram 429,8% entre 2018 e 2025, e o meio eletrônico respondeu por 15,3% das 2.261.055 ocorrências de 2025 (346.753), enquanto 15,8% da população de 16 anos ou mais declarou ter sido vítima de golpe online no último ano. No CERT.br, 94,96% das notificações de fraude de 2025 foram de *phishing*. A convergência é com o engano do usuário como mecanismo dominante, e não com a invasão técnica de dispositivos. As séries, no entanto, medem coisas distintas (notificações técnicas voluntárias, boletins de ocorrência e pesquisas de vitimização) e não devem ser somadas nem comparadas diretamente.

**PE2 – comportamento.** A literatura e as pesquisas descrevem um padrão consistente: preocupação com a segurança convivendo com proteção incompleta, confiança em medidas parciais e dificuldade de distinguir redes falsas (KLASNJA et al., 2009; MAIMON et al., 2017; SOMBATRUANG; SASSE; BADDELEY, 2016; CHOI; CARPENTER; KO, 2022; PANDA SECURITY, 2025). O estudo de campo sobre alertas de navegador indica que o desenho da interface pode levar a maioria a prosseguir diante de um alerta (AKHAWE; FELT, 2013). Para o material educativo, a implicação é treinar a decisão no momento do risco (confirmar o nome da rede, recusar o pedido de senha, não ignorar alertas) e não apenas informar.

**PE3 – estimativas.** As estimativas aritméticas são robustas à incerteza dos dados de entrada (base de cerca de 166 milhões de pessoas de 16 anos ou mais, com variação de ±0,5%), mas a exposição ao Wi-Fi aberto depende de uma premissa fraca, que varia de 16,6 a 83,2 milhões de pessoas conforme o cenário. A conclusão defensável é de ordem de grandeza (dezenas de milhões), e não de valor pontual.

**PE4 – cobertura do material.** Os onze temas de risco e proteção mapeados aparecem em pelo menos três dos sete objetos, de modo que nenhum tema está totalmente descoberto. A menor cobertura é a de *Man-in-the-Middle* e interceptação de tráfego (9 menções, em três objetos), e o quiz não avalia esse tema, a conexão automática, as atualizações nem o encerramento do uso, que só aparecem no teste final. Os itens de avaliação têm uma pista de construção (a alternativa correta é a mais longa nos 15 itens), e o pré e o pós-teste previstos compartilham apenas um tema, o que compromete a comparação entre eles.

**PE5 – teste com pessoas.** O plano de 8 a 15 participantes só detecta efeitos grandes; para um efeito moderado, são necessários cerca de 34 participantes. Além do tamanho amostral, o delineamento sem grupo de controle e com itens diferentes no pré e no pós limita a interpretação causal, e os itens do instrumento precisam de revisão de construção (seção 4.5). O caderno deixa pronto o procedimento de análise (diferença média, *bootstrap*, Wilcoxon e $d_z$), que roda automaticamente quando houver dados reais.

**Limitações do estudo.** (i) Os dados são secundários e, em parte, lidos de reportagens que citam os documentos originais; (ii) nenhuma fonte mede golpes originados em Wi-Fi público no Brasil; (iii) as pesquisas de fornecedores de segurança têm interesse comercial, datas e países diferentes e, no caso do Brasil, não informam a amostra; (iv) a data da pesquisa de vitimização do FBSP diverge entre reportagens; (v) a análise de cobertura temática por expressões regulares mede presença, e não profundidade ou correção; e (vi) o teste com pessoas ainda não foi aplicado, de modo que nenhuma conclusão sobre efetividade do material é feita.

<h2 align="center">APÊNDICE B – Instrumento do teste com pessoas</h2>

**Texto de abertura (consentimento).** “Esta pesquisa é voluntária e anônima. Não pedimos seu nome nem e-mail. **Parte D – Pós-teste.** Itens 1, 2, 6, 8 e 10 do teste final (listados abaixo).

**Parte E – Percepção.** (1) O material foi claro? (1 a 5) (2) O material é útil no seu dia a dia? (1 a 5) (3) Você compartilharia este material com alguém? (sim / talvez / não) (4) O que você achou confuso ou mudaria? (aberta, opcional)

````python
PRE = [("Quiz", i) for i in range(1, 6)]
POS = [("Teste final", i) for i in (1, 2, 6, 8, 10)]
def listar(titulo, ids):
    linhas = [f"**{titulo}**\n"]
    for k, (inst, i) in enumerate(ids, 1):
        r = it_df[(it_df.instrumento == inst) & (it_df["item"] == i)].iloc[0]
        alts = r.alternativas.replace(" | ", "<br>")
        linhas.append(f"{k}. {r.enunciado}<br>{alts}<br>*Resposta correta: {r.correta}. Tema principal: {r.tema}.*\n")
    display(Markdown("\n".join(linhas)))
listar("Pré-teste (itens 1 a 5 do quiz)", PRE)
listar("Pós-teste (itens 1, 2, 6, 8 e 10 do teste final)", POS)
tp = [it_df[(it_df.instrumento == a) & (it_df["item"] == b)].iloc[0].tema for a, b in PRE]
tq = [it_df[(it_df.instrumento == a) & (it_df["item"] == b)].iloc[0].tema for a, b in POS]
comuns = sorted(set(tp) & set(tq))
display(Markdown(f"Temas comuns ao pré e ao pós: **{', '.join(comuns) if comuns else 'nenhum'}**. "
                 f"Temas só no pré: {', '.join(sorted(set(tp) - set(tq))) or 'nenhum'}. Temas só no pós: {', '.join(sorted(set(tq) - set(tp))) or 'nenhum'}."))
````
